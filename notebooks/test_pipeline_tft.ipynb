## Prepare Data

```python
from eco2mix_prepare_tft_data import Eco2mixPrepareTFTData

splitter = Eco2mixPrepareTFTData(
    input_path="../data/06_features/eco2mix/minute/tempo_cleaned_consumption_added_features_data.csv",
    output_dir="../data/07_model_input/eco2mix/tft/minute/data",
    series_id_value="France",
    columns_to_drop=["Prévision J", "Prévision J-1"]
)

splitter.run()
```

## Create TS Dataset

```python
from eco2mix_create_tsdataset_tft import Eco2mixCreateTSDatasetsTFT

creator = Eco2mixCreateTSDatasetsTFT(
    df_path="../data/07_model_input/eco2mix/tft/minute/data/df.parquet",
    output_dir="../data/07_model_input/eco2mix/tft/minute/tsdataset",
    test_cutoff_years = 1,
    val_duration_years = 1,
    max_encoder_length=6*48,
    max_prediction_length=48,
    known_reals=["time_idx", "apparent_temperature", "is_vacation"],
    unknown_reals=["Consommation", "sunshine_duration"],
    known_categoricals=["weekday", "month", "season"],
    unknown_categoricals=[],
    static_reals=[],
    static_categoricals=[]
)

creator.run()
```

## Train model

```python
from eco2mix_train_model_tft import Eco2mixTrainModelTFT

trainer = Eco2mixTrainModelTFT(
    training_dataset_path="../data/07_model_input/eco2mix/tft/minute/tsdataset/tft_training_dataset.pkl",
    validation_dataset_path="../data/07_model_input/eco2mix/tft/minute/tsdataset/tft_validation_dataset.pkl",
    params_path="../conf/experiments/tft.yml",
    output_root="../data/09_models/tft/minute/staging"
)

trainer.run()
```

## Evaluate model

```python
from eco2mix_evaluate_model_tft import Eco2mixEvaluateModelTest

evaluator = Eco2mixEvaluateModelTest(
    model_path="../data/09_models/tft/minute/staging/20250505_212911/model.ckpt",
    test_dataset_path="../data/07_model_input/eco2mix/tft/minute/tsdataset/tft_test_dataset.pkl",
    output_dir="../data/09_models/tft/minute/staging/20250505_212911"
)

evaluator.run()
```
